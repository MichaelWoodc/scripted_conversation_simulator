# Generate base audio notebooks from DOCX files

Run all cells in **this generator notebook**. It creates one combined notebook and
one notebook per Word file. No speech model is loaded and no audio is generated here.

The generated notebooks contain raw answer text in code cells. Running an answer
cell generates its audio and shows **Play preview** and **Save and replace WAV**.
Your existing working notebooks are separate from the default output folder.


```python
from pathlib import Path
import os
import sys

project = next((p for p in (Path.cwd(), *Path.cwd().parents)
                if (p / "bin" / "notebook_generator.py").is_file()), None)
if project is None:
    raise RuntimeError("Open this notebook from the TTS project folder.")
os.chdir(project)
if str(project / 'bin') not in sys.path:
    sys.path.insert(0, str(project / 'bin'))

from notebook_generator import generate_notebooks
from IPython.display import display, FileLink
print("Ready to build notebooks from Word files.")
```

```python
# Change these folders if needed.
docx_folder = project / "data" / "scripts"
output_folder = project / "notebooks" / "generated"

voice = "af_heart"
speed = 1.0
preserve_edits = True  # Keep text/code edits in previously generated notebooks.
```

```python
results = generate_notebooks(
    docx_folder=docx_folder,
    output_folder=output_folder,
    project_root=project,
    voice=voice,
    speed=speed,
    preserve_edits=preserve_edits,
)
for item in results:
    print(f"{item['answers']} answers; {item['preserved_cells']} existing cells preserved")
    display(FileLink(str(item["path"].relative_to(project))))
print("Done. Open a generated notebook, run its setup, then run the answer cell you want.")
```

The source Word files must have QUESTION/ANSWER labels. Only answers are narrated.
Set `preserve_edits = False` to rebuild generated notebooks entirely from the Word text.
In a generated notebook, **Save and replace WAV** writes the preview to its matching
numbered file under `dist/media/edited`.
