# Answer audio from Word scripts

Run the setup cell once. Edit the raw `text`, `voice`, and `speed` in an answer cell, then run that cell to generate its audio preview. Click **Play preview**, then **Save and replace WAV** when happy.

Save replaces the matching numbered file in `dist/media/edited/<voice>/speed_1p00/<Word filename>/`. The exact destination appears under the player. No audio effects are added.

**Ctrl+S** saves your text inside this notebook. Run only the answer cells you want to generate.

```python
from pathlib import Path
import os
import sys
from IPython.display import display

project = next((p for p in (Path.cwd(), *Path.cwd().parents)
                if (p / 'bin' / 'cell_answer_review.py').is_file()), None)
if project is None:
    raise RuntimeError('Open this notebook inside the TTS project folder.')
os.chdir(project)
if str(project / 'bin') not in sys.path:
    sys.path.insert(0, str(project / 'bin'))

from answer_tts import load_answers
from cell_answer_review import CellAnswerStudio
answers = load_answers(project / 'data' / 'scripts')
if 'cell_studio' not in globals() or not isinstance(cell_studio, CellAnswerStudio):
    cell_studio = CellAnswerStudio(project_root=project, answers=answers)
else:
    cell_studio.answers = answers
display(cell_studio.playback.widget)
display(cell_studio.export_widget)
print('Ready. Run an answer cell to generate its audio, then Play or Save below it.')
```
