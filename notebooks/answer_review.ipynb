# Edit the answer text in code, then Play and Save

1. Run the **setup cell** below once.
2. Edit `text = """..."""` in the answer's code cell. Keep or change its `voice` and `speed`.
3. Run **that answer cell** to generate just its audio.
4. Click **Play preview**, then **Save and replace WAV** when happy.

Save replaces the matching numbered WAV in
`dist/media/edited/<voice>/speed_1p00/<Word filename>/`.
The exact destination is shown below each player. The matching `.txt` and `.json`
are replaced too. Running a cell generates a preview; Save writes the WAV.

Your wording, punctuation, and line breaks go straight to the model. No audio effects are applied.
Choose your **Speaker** and use **Test speaker** to check playback.

Use **Ctrl+S** to save your raw text in this notebook. Running an answer cell also backs up its
text in `data/answer_edits.json`. Save the notebook before using **Export for another PC**.
Run the setup once, then only the answer cells you want to generate.


```python
from pathlib import Path
import os
import sys
from IPython.display import display

project = next((p for p in (Path.cwd(), *Path.cwd().parents)
                if (p / 'bin' / 'cell_answer_review.py').is_file()), None)
if project is None:
    raise RuntimeError('Open this notebook from the project folder.')
os.chdir(project)
if str(project / 'bin') not in sys.path:
    sys.path.insert(0, str(project / 'bin'))

from review_bootstrap import get_cell_studio
cell_studio = get_cell_studio(globals())
display(cell_studio.playback.widget)
display(cell_studio.export_widget)
print('Ready. Edit text in an answer cell, then run that cell for Play / Save controls.')
print('Save folder:', project / 'dist' / 'media' / 'edited')
```
